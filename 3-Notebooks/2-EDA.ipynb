```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
campaign = pd.read_csv('../Healthcare analysis/clean_campaign.csv')
audience = pd.read_csv('../Healthcare analysis/clean_audience.csv')
spend = pd.read_csv('../Healthcare analysis/clean_spend.csv')
conversion = pd.read_csv('../Healthcare analysis/clean_conversion.csv')
```

Campaign Distribution

```python
campaign['Campaign_Name'].value_counts()
```

Channel Distribution

```python
campaign['Channel'].value_counts()
```

Spend by Channel

```python
campaign.groupby('Channel')['Spend'].sum().sort_values()
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    data=campaign,
    x='Channel',
    y='Spend')

plt.title('Campaign Spend by Channel')
plt.show()
```
