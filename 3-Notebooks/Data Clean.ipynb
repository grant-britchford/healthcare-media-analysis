```python
import pandas as pd
import numpy as np
```

```python
audience = pd.read_csv('../Healthcare analysis/Audience_Data.csv')
campaign = pd.read_csv('../Healthcare analysis/Campaign_Performance.csv')
conversion = pd.read_csv('../Healthcare analysis/Conversions.csv')
spend = pd.read_csv('../Healthcare analysis/Media_Spend.csv')
```

Standardise Campaign Names

```python
campaign['Campaign_Name'] = campaign['Campaign_Name'].replace({'Diabtes Plus':'Diabetes Plus', 'Cancer Care':'CancerCare'})
```

Standardise Channels

```python
campaign['Channel'] = campaign['Channel'].replace({'Goggle':'Google', 'Youtube':'YouTube'})

spend['Channel'] = spend['Channel'].replace({'LinkdIn':'LinkedIn', 'Goggle':'Google', 'Youtube':'YouTube'})
```

Date Conversion

```python
campaign['Campaign_Date'] = pd.to_datetime(campaign['Campaign_Date'], 'dayfirst=True')

spend['Spend_Date'] = pd.to_datetime(spend['Spend_Date'],'dayfirst=True')
```

Missing Impressions

```python
campaign['Impressions'] = campaign['Impressions'].fillna(campaign['Impressions'].median())
```

Missing Clicks

```python
campaign['Clicks'] = campaign['Clicks'].fillna(campaign['Clicks'].median())
```

Age Clean

```python
audience['Age'] = pd.to_numeric(audience['Age'], errors='coerce')

audience['Age'] = audience['Age'].fillna(audience['Age'].median())
```

Condition Clean

```python
audience['Condition'] = audience['Condition'].fillna('Unknown')
```

Duplicate Clean

```python
duplicates = campaign.duplicated()
duplicates.sum()
```

```python
duplicates = spend.duplicated()
duplicates.sum()
```

```python
duplicates = conversion.duplicated()
duplicates.sum()
```

```python
duplicates = audience.duplicated()
duplicates.sum()
```

Clean Verification

```python
campaign = campaign.drop_duplicates()
```

```python
duplicates = campaign.duplicated()
duplicates.sum()
```

Spend Clean

```python
spend['Spend_Date'].isnull().sum()
```

```python
spend[spend['Spend_Date'].isnull()]
```

```python
spend['Spend_Date'] = pd.to_datetime(spend['Spend_Date'], 'dayfirst=True')
```

```python
spend['Spend_Date'].isnull().sum()
```

```python
spend = spend.dropna(subset=['Spend_Date'])
```

```python
spend['Spend_Date'].isnull().sum()
```

```python
duplicates = spend['Spend_Date'].duplicated()
duplicates.sum()
```

```python
spend = spend['Spend_Date'].drop_duplicates()
```

Processed Data Save

```python
campaign.to_csv('../Healthcare analysis/clean_campaign.csv', index=False)

audience.to_csv('../Healthcare analysis/clean_audience.csv', index=False)

spend.to_csv('../Healthcare analysis/clean_spend.csv', index=False)

conversion.to_csv('../Healthcare analysis/clean_conversion.csv', index=False)
```
