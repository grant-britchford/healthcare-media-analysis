```python
import pandas as pd
import numpy as np
```

```python
conversion = pd.read_csv('../Healthcare analysis/clean_conversion.csv')
audience = pd.read_csv('../Healthcare analysis/clean_audience.csv')
spend = pd.read_csv('../Healthcare analysis/clean_spend.csv')
campaign = pd.read_csv('../Healthcare analysis/clean_campaign.csv')
```

Merge Datasets

```python
merged = conversion.merge(campaign, on='Campaign_ID', how='left')

merged = merged.merge(audience, on='Patient_ID', how='left')
```

Define Age Groups

```python
merged['Age Group'] = pd.cut(
    merged['Age'],
    bins=[0,25,45,65,100],
    labels=['18-25', '26-45', '46-65', '65+'])
```

Conversion Flag

```python
merged['Conversion_Flag'] = np.where(
    merged['Prescription_Conversion']=='Yes',
    1,
    0)
```

CTR Claculation

```python
merged['CTR'] = (merged['Clicks'] / merged['Impressions']) * 100
```

Revenue by Campaign

```python
Campaign_Revenue = merged.groupby(
    'Campaign_Name')['Revenue'].sum()
```

Conversion Rate by Campaign

```python
Campaign_Conversion = merged.groupby(
    'Campaign_Name')['Conversion_Flag'].mean() * 100
```

Dashboard CSV 

```python
merged.to_csv('../Healthcare analysis/Dashboard_Data.csv', index=False)
```
